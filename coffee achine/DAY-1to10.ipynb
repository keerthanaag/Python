### Caesar cipher

```python
lst=[chr(97+i) for i in range(0,26) ]
print(lst)
d={chr(97+i):i for i in range(26)}
print(d)
```

```python
shift=4
```

```python
word='abraz'
```

```python
def encode(word,shift):
    temp=''
    for x in word:
        val=d.get(x)
        val+=shift
        if val>25:
            val-=26
        temp+=lst[val]
    print(temp)
```

```python
def decode(temp,shift):
    word=''
    for x in temp:
        val=d.get(x)
        val-=shift
        if val<0:
            val+=26
        word+=lst[val]
    print(word)
```
