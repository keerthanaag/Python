```python
import numpy as np
MENU = {
    "espresso": {
        "ingredients": {
            "water": 50,
            "coffee": 18,
        },
        "cost": 1.5,
    },
    "latte": {
        "ingredients": {
            "water": 200,
            "milk": 150,
            "coffee": 24,
        },
        "cost": 2.5,
    },
    "cappuccino": {
        "ingredients": {
            "water": 250,
            "milk": 100,
            "coffee": 24,
        },
        "cost": 3.0,
    }
}

resources = {
    "water": 300,
    "milk": 200,
    "coffee": 100,
    "money" : 0,
}
```

```python
def check_quantity(preference):
    flag=0
    for item in MENU[preference]["ingredients"]:
        #print(item)
        if MENU[preference]["ingredients"][item] >= resources[item]:
            print(f"Sorry there is not enough {item}.")
            flag=1
            break
    return flag
```

```python

def process_coins(preference):
    quaters,dimes,nickles,pennies = 0.25,0.10,0.05,0.01
    print(f"The cost of {preference} is ${MENU[preference]['cost']}")
    quaters_user=int(input("Enter the quaters: "))
    dimes_user= int(input("enter the dimes: "))
    nickles_user = int(input("enter the nickles: "))
    pennies_user = int(input("enter the pennies: "))
    cal=(quaters*quaters_user)+(nickles*nickles_user)+(dimes*dimes_user)+(pennies*pennies_user)
    if cal < MENU[preference]["cost"]:
        return "Insufficient Amount."
    else:
        balance = cal - MENU[preference]["cost"]
        val=resources["money"]+MENU[preference]["cost"]
        resources.update({"money":val})
        return f"The balance amount ${np.round(balance,2)} has been returned.\nThank You for purchasing"
```

```python
def update_quantity(preference):
    for item in MENU[preference]["ingredients"]:
        value=resources[item]
        used= MENU[preference]["ingredients"][item]
        ans = value -used
        resources.update({item:ans})
```

```python
def ask():
    while(1):
        interest=input('What would you like? (espresso/latte/cappuccino): ')
        if interest.lower() == "report":
            print("Resources\n")
            for x in resources:
                print(f"{x} : {resources[x]}")
        elif interest.lower() == "off":
            break
        else:
            ans= check_quantity(interest.lower())
            if ans == 1:
                print("thank you.")
                break
            money = process_coins(interest.lower())
            update_quantity(interest.lower())
            print(money)
        
```

```python
ask()
```
